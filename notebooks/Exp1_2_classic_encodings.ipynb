```python
from transformers.pipelines.pt_utils import KeyDataset
from pathlib import Path

from datasets import load_dataset


from mhcicleave.encodings import blosum62_batch,physic_batch

outdir = Path("../data/encodings/")
if not outdir.is_dir():
    outdir.mkdir()
```

```python
downsampled_dataset = load_dataset("csv",data_files="../data/c_test_downs_exp1.csv", split="train")
downsampled_dataset.set_format("numpy")
```

```python

batch_size = 64

# Apply the map function with batching
encoded_blosum62 = downsampled_dataset.map(
    blosum62_batch,
    batched=True,  # Process in batches
    batch_size=batch_size,
    remove_columns=['cleavage_window'],
    num_proc=10
)

# by default properties 1 corresponding to Meiler et al features
encoded_physic = downsampled_dataset.map(
    physic_batch,
    batched=True,  # Process in batches
    batch_size=batch_size,
    remove_columns=['cleavage_window'],
    num_proc=10
)
```

```python
encoded_blosum62.save_to_disk(str(outdir/"blosum62"))
encoded_physic.save_to_disk(str(outdir/"physic"))
```
