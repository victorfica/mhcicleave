```python
import requests
import pandas as pd

# API endpoint for querying ligand data
url = 'https://query-api.iedb.org/api/v1/mhcligand/'

# Define the query parameters
query = {
    "filters": [
        {"field": "mhc_class", "operator": "=", "value": "I"},
        {"field": "assay_type", "operator": "contains", "value": "ligand elution"},
        {"field": "host_organism_name", "operator": "=", "value": "Mammalia"}
    ],
    "limit": 10000  # Adjust the limit based on the size of the data you need
}

# Make the POST request to the API
response = requests.post(url, json=query)

# Check if the request was successful
if response.status_code == 200:
    data = response.json()
    df = pd.json_normalize(data['data'])
    
    # Save the data to a CSV file
    df.to_csv('iedb_mhc_class_i_ligands.csv', index=False)
    print('Data downloaded and saved as iedb_mhc_class_i_ligands.csv')
    
    # Display the first few rows
    print(df.head())
else:
    print(f'Failed to download data. Status code: {response.status_code}')
    print(response.text)
```
